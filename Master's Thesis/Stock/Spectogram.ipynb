```python
# Kütüphaneler ve işleme araçları
import matplotlib.pyplot as plt
import numpy as np
import librosa
import librosa.display

# Ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav"
file_2 = "/Users/nazimerensahin/Desktop/2.wav"

# Ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1"
label_2 = "Ses Dosyası 2"

# STFT (Short-Time Fourier Transform) parametreleri
n_fft = 2048
hop_length = 512
cmap = 'magma'

# Grafik oluşturma
plt.figure(figsize=(12, 8))

# Birinci ses dosyasının analizi
y1, sr1 = librosa.load(file_1, sr=None)
D1 = librosa.stft(y1, n_fft=n_fft, 
                  hop_length=hop_length, window='hann')
S_db1 = librosa.amplitude_to_db(np.abs(D1), ref=np.max)

# İkinci ses dosyasının analizi
y2, sr2 = librosa.load(file_2, sr=None)
D2 = librosa.stft(y2, n_fft=n_fft, 
                  hop_length=hop_length, window='hann')
S_db2 = librosa.amplitude_to_db(np.abs(D2), ref=np.max)


# Birinci spektrogram grafiği
plt.subplot(2, 1, 1)
librosa.display.specshow(S_db1, sr=sr1, 
hop_length=hop_length, x_axis='time', 
y_axis='log', cmap=cmap) 
plt.colorbar(format='%+2.0f dB')
plt.title(f"{label_1} - Spektrogram")
plt.xlabel("Zaman (saniye)")
plt.ylabel("Frekans (Hz)")

# İkinci spektrogram grafiği
plt.subplot(2, 1, 2)
librosa.display.specshow(S_db2, sr=sr2, 
hop_length=hop_length, x_axis='time', 
y_axis='log', cmap=cmap) 
plt.colorbar(format='%+2.0f dB')
plt.title(f"{label_2} - Spektrogram")
plt.xlabel("Zaman (saniye)")
plt.ylabel("Frekans (Hz)")

# Genel başlık ve gösterim
plt.tight_layout(rect=[0, 0, 1, 0.95]) 
plt.suptitle("İki Model Arası Spektrogram Karşılaştırması",
             fontsize=16)
plt.show()
```
