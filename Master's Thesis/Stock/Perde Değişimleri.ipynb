```python
import matplotlib.pyplot as plt # Grafik çizimleri için kullanılan kütüphane
import numpy as np # Sayısal hesaplamaları
import librosa # Ses yükleme ve analizi
import librosa.display # Ses dalga formu görselleştirme

# Ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav" # İlk karşılaştırılacak ses dosyasının yolu
file_2 = "/Users/nazimerensahin/Desktop/2.wav" # İkinci karşılaştırılacak ses dosyasının yolu

# Ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1" # Grafik etiketi için ilk dosya ismi
label_2 = "Ses Dosyası 2" # Grafik etiketi için ikinci dosya ismi

# pYIN (pitch tracking) ayarları
fmin = librosa.note_to_hz("C2") # Minimum analiz edilecek frekans (Hz)
fmax = librosa.note_to_hz("C7") # Maksimum analiz edilecek frekans (Hz)

# Birinci ses dosyasını yükle ve pitch analizini yapma
sound_1, sr_1 = librosa.load(file_1, sr=None) # İlk dosyayı orijinal örnekleme oranıyla yükle
pitch_1, _, _ = librosa.pyin(sound_1, fmin=fmin, fmax=fmax, sr=sr_1) # pYIN ile pitch analizi
pitch_time_1 = librosa.times_like(pitch_1, sr=sr_1) # Zaman ekseni oluştur

# İkinci ses dosyasını yükle ve pitch analizini yapma
sound_2, sr_2 = librosa.load(file_2, sr=None) # İkinci dosyayı yükleme
pitch_2, _, _ = librosa.pyin(sound_2, fmin=fmin, fmax=fmax, sr=sr_2) # Perde analizi
pitch_time_2 = librosa.times_like(pitch_2, sr=sr_2) # Zaman ekseni oluşturma

# Grafik oluşturma
plt.figure(figsize=(12, 5)) # Grafik boyutunu ayarlama
plt.plot(pitch_time_1, pitch_1, label=label_1, color="blue", linewidth=2) # Birinci dosyanın perde eğrisi
plt.plot(pitch_time_2, pitch_2, label=label_2, color="red", linewidth=2) # İkinci dosyanın perde eğrisi
plt.title("Pitch (Perde) Karşılaştırması") # Grafik başlığı
plt.xlabel("Zaman (saniye)") # X ekseni etiketi
plt.ylabel("Frekans (Hz)") # Y ekseni etiketi
plt.legend(loc="upper right") # Etiketleri sağ üstte gösterme
plt.grid(True) # Izgara çizgileri ekleme
plt.tight_layout() # Otomatik yerleşim
plt.show() # Grafiği gösterme
```
