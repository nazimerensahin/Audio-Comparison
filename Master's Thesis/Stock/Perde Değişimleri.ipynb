```python
# Gerekli kütüphanelerin içe aktarılması
import matplotlib.pyplot as plt
import numpy as np
import librosa
import librosa.display

# İşlenecek ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav"
file_2 = "/Users/nazimerensahin/Desktop/2.wav"

# İşlenecek ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1"
label_2 = "Ses Dosyası 2"

# pYIN algoritması için minimum ve maksimum perde frekanslarını ayarla
fmin = librosa.note_to_hz("C2")
fmax = librosa.note_to_hz("C7")

# İlk ses dosyasını yükle ve perde (pitch) bilgisini hesapla
sound_1, sr_1 = librosa.load(file_1, sr=None)
pitch_1, _, _ = librosa.pyin(sound_1, fmin=fmin, fmax=fmax, sr=sr_1)
pitch_time_1 = librosa.times_like(pitch_1, sr=sr_1)

# İkinci ses dosyasını yükle ve perde (pitch) bilgisini hesapla
sound_2, sr_2 = librosa.load(file_2, sr=None)
pitch_2, _, _ = librosa.pyin(sound_2, fmin=fmin, fmax=fmax, sr=sr_2)
pitch_time_2 = librosa.times_like(pitch_2, sr=sr_2)

# Grafik penceresini oluştur
plt.figure(figsize=(12, 5))

# Her iki ses dosyasının pitch (perde) bilgisini zaman eksenine göre çiz
plt.plot(pitch_time_1, pitch_1, label=label_1, color="blue", linewidth=2)
plt.plot(pitch_time_2, pitch_2, label=label_2, color="red", linewidth=2)

# Grafik başlığını ve eksen etiketlerini ayarla
plt.title("Pitch (Perde) Karşılaştırması")
plt.xlabel("Zaman (saniye)")
plt.ylabel("Frekans (Hz)")

# Görsel ayarlamalar
plt.legend(loc="upper right")
plt.grid(True)
plt.tight_layout()
plt.show()
```
