```python
# Gerekli kütüphanelerin içe aktarılması
import matplotlib.pyplot as plt
import numpy as np
import librosa
import librosa.display

# İşlenecek ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav"
file_2 = "/Users/nazimerensahin/Desktop/2.wav"

# İşlenecek ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1"
label_2 = "Ses Dosyası 2"

# Ses dosyalarını yükle
wave_1, sr_1 = librosa.load(file_1, sr=None)
wave_2, sr_2 = librosa.load(file_2, sr=None)

# FFT ile frekans ve genlik bilgilerini hesapla
def compute_fft(signal, sr):
	fft = np.fft.fft(signal)
	magnitude = np.abs(fft[:len(fft)//2])
	freqs = np.fft.fftfreq(len(fft), 1/sr)[:len(fft)//2]
	return freqs, magnitude

# Her iki ses dosyası için frekans spektrumu hesapla
freqs1, mag1 = compute_fft(wave_1, sr_1)
freqs2, mag2 = compute_fft(wave_2, sr_2)

# Grafik boyutunu ayarla
plt.figure(figsize=(12, 6))

# Frekans spektrumlarını çiz
plt.plot(freqs1, mag1, label="Ses Dosyası 1", 
color="blue", alpha=0.7)
plt.plot(freqs2, mag2, label="Ses Dosyası 2", 
color="red", alpha=0.7)

# Görsel ayarlamalar
plt.title("Frekans Spektrumu")
plt.xlabel("Frekans (Hz)")
plt.ylabel("Genlik")
plt.legend()
plt.grid(True)
plt.xlim(0, 20000)

# Düzenlemeleri uygula ve göster
plt.tight_layout()
plt.show()
```
