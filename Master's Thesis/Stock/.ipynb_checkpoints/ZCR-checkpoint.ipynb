```python
# Kütüphaneler ve işleme araçları
import matplotlib.pyplot as plt
import librosa
import librosa.display

# Ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav"
file_2 = "/Users/nazimerensahin/Desktop/2.wav"

# Ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1"
label_2 = "Ses Dosyası 2"

# Ses dosyalarını yükleme
# y = ses sinyali (dalga formu), sr = örnekleme hızı
y1, sr1 = librosa.load(file_1, sr=None)
y2, sr2 = librosa.load(file_2, sr=None)

# Zero-Crossing Rate hesaplama
# Her ses sinyali için ZCR değerini çerçeve bazında hesaplama
zcr1 = librosa.feature.zero_crossing_rate(y1)[0]
zcr2 = librosa.feature.zero_crossing_rate(y2)[0]

# Zaman eksenlerini oluşturma
# ZCR değerlerine karşılık gelen zaman vektörlerini hesaplama
time1 = librosa.times_like(zcr1, sr=sr1)
time2 = librosa.times_like(zcr2, sr=sr2)

# Grafik çizimi
plt.figure(figsize=(12, 6))
plt.plot(time1, zcr1, label=label_1, color="blue")
plt.plot(time2, zcr2, label=label_2, color="red")

plt.title("Zero-Crossing Rate Karşılaştırması")
plt.xlabel("Zaman (saniye)")
plt.ylabel("ZCR (0-1 aralığı)")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
