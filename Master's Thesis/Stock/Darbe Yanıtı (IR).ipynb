```python
# Kütüphaneler ve işleme araçları
import matplotlib.pyplot as plt
import numpy as np
import librosa

# Ses dosyalarının yolları
file_1 = "/Users/nazimerensahin/Desktop/1.wav"
file_2 = "/Users/nazimerensahin/Desktop/2.wav"

# Ses dosyalarının isimlendirmeleri
label_1 = "Ses Dosyası 1"
label_2 = "Ses Dosyası 2"

# Ses dosyalarını yükleme
wave_1, sr_1 = librosa.load(file_1, sr=None)
wave_2, sr_2 = librosa.load(file_2, sr=None)

# Zaman eksenlerini oluşturma
time_1 = np.linspace(0, len(wave_1) / sr_1, len(wave_1))
time_2 = np.linspace(0, len(wave_2) / sr_2, len(wave_2))

# Grafik çizimi
plt.figure(figsize=(12, 4))

# Her iki ses dosyasının darbeye verdiği yanıtın çizimi
tl = plt.plot(time_1, wave_1, label=label_1, alpha=0.8, color="red")
tr = plt.plot(time_2, wave_2, label=label_2, alpha=0.8, color="blue")

# Grafik ayarları
plt.title(f"Darbe Yanıtı Karşılaştırması: {label_1} vs {label_2}")
plt.xlabel("Zaman (saniye)")
plt.ylabel("Genlik")
plt.legend(loc="lower left")
plt.grid(True)
plt.tight_layout()
plt.show()
```
